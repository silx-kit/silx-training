## Introduction

We have the **daily mean temperature** (in °C) measured in Grenoble for every day from 1 January 1950 to 31 December 2025.

To get a regular 2D array, 29 February has been removed from leap years, so every year has exactly 365 days.

The temperatures are stored as a 2D dataset of shape (76, 365) in the `temperatures.npy` file:

* each **row** is one year (1950, 1951, ..., 2025),
* each **column** is one day of the year (1 January, 2 January, ..., 31 December).

The list of years is stored in the `years.txt` file.

The question is: **has the climate in Grenoble changed, and by how much?**

1. Look at the data and get a feeling of it

```python
# Part I

import numpy as np

# Load data
temperatures = np.load("temperatures.npy")
years = np.loadtxt("years.txt", dtype=int)

# Part II

# Typical temperature of each day of the year
temperatures_per_day = np.mean(temperatures, axis=0)

# Mean temperature of each year
temperatures_per_year = np.mean(temperatures, axis=1)

# Warmest and coldest years
warmest_year = years[np.argmax(temperatures_per_year)]
coldest_year = years[np.argmin(temperatures_per_year)]
warm_years = years[temperatures_per_year > 13]

# Part III

# Separate the reference period from the recent period
# 1. With a condition over years
reference = temperatures[(years >= 1951) & (years <= 1980)]
recent = temperatures[years >= 1996]
# 2. With slicing
reference = temperatures[1:31]
recent = temperatures[-30:]

# Average temperature of each day for both periods
reference_mean = np.mean(reference, axis=0)
recent_mean = np.mean(recent, axis=0)

# Warming
warming = recent_mean - reference_mean
warming_mean = np.mean(warming)

# Bonus: hot days
temperatures_max = np.load("temperatures_max.npy")
hot_days_per_year = np.sum(temperatures_max > 30, axis=1)
```
